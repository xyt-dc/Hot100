```python
def merge_sort(nums):
    if len(nums) == 1:
        return nums
    mid = len(nums)>>1
    left_nums = merge_sort(nums[:mid])
    right_nums = merge_sort(nums[mid:])

    res = []
    i,j=0,0
    while i < len(left_nums) or j < len(right_nums):
        if i >= len(left_nums):
            res.append(right_nums[j])
            j+=1
        elif j >= len(right_nums):
            res.append(left_nums[i])
            i+=1
        elif left_nums[i]<right_nums[j]:
            res.append(left_nums[i])
            i+=1
        else:
            res.append(right_nums[j])
            j+=1

    return res
```

```python

class Solution:
    #最大值到栈顶
    def heapinsert(self,arr:list):
        if len(arr) < 2:
            return arr
        for i in range(len(arr)):
            parent= (i-1)//2
            while i > 0 and arr[i] > arr[parent] :
                arr[i],arr[parent] = arr[parent],arr[i]
                i = parent
                parent= (i-1)//2
        return arr
    #弹出最大值并让其再次成为堆再
    def heapPop(self,arr:list) -> int:
        ans = arr[0]
        last = arr.pop()
        if arr:
            arr[0] = last
            i = 0
            left = 2*i + 1
            while left < len(arr):
                larger = left if left + 1 >= len(arr) or arr[left] > arr[left+1] else left+1
                larger = i if arr[i] > arr[larger] else larger
                if i == larger:
                    break
                arr[i],arr[larger] = arr[larger],arr[i]
                i = larger
                left = 2*i +1
        return ans
    #碰撞后产生的新石头再加入堆
    def heapPush(self,arr:list,nums:int):
        arr.append(nums)
        i = len(arr) - 1
        parent= (i-1)//2
        while i > 0 and arr[i] > arr[parent] :
            arr[i],arr[parent] = arr[parent],arr[i]
            i = parent
            parent= (i-1)//2


    def lastStoneWeight(self, stones: list[int]) -> int:
        arr = self.heapinsert(stones)
        if len (arr) ==1:
            return arr[0]
        if len (arr) ==0:
            return 0
        while len(arr) > 1:
            a = self.heapPop(arr)
            b = self.heapPop(arr)
            nums = a-b if a> b else b-a
            if nums > 0:
                self.heapPush(arr,nums)
        return arr[0] if len(arr) == 1 else 0



```

```python
#创建树的功能函数
def build_tree(nums):
    if not nums or nums[0] is None:
        return None

    # 创建根节点
    head = TreeNode(nums[0])

    # 队列保存等待添加孩子的节点
    queue = deque([head])

    i = 1

    while queue and i < len(nums):
        # 取出当前父节点
        cur = queue.popleft()

        # 创建左孩子
        if i < len(nums) and nums[i] is not None:
            cur.left = TreeNode(nums[i])
            queue.append(cur.left)

        i += 1

        # 创建右孩子
        if i < len(nums) and nums[i] is not None:
            cur.right = TreeNode(nums[i])
            queue.append(cur.right)

        i += 1

    return head
```

```python
#leecode  102  层排序
from collections import deque
class TreeNode:
         def __init__(self, val=0, left=None, right=None):
            self.val = val
            self.left = left
            self.right = right

class Solution:
    def levelOrder(self, root: TreeNode | None) -> list[list[int]]:
        ans = []

        if root is None:
            return ans

        queue = deque()
        queue.append(root)

        while queue:
            # 当前层有多少个节点
            currentLevelSize = len(queue)

            # 创建一个新的空列表，保存当前层
            ans.append([])

            # 只处理当前层的节点
            for i in range(currentLevelSize):
                node = queue.popleft()

                # 将当前节点放入最后一层的列表
                ans[-1].append(node.val)

                if node.left:
                    queue.append(node.left)

                if node.right:
                    queue.append(node.right)

        return ans
root = build_tree([1,2,3,None,5,6,7])
ans = Solution().levelOrder(root)
print(ans)
```
